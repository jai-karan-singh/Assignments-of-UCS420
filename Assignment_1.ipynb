```python
n = int(input("Enter a number to find all numbers divisible by 7 and 9 from 1 to n: "))
total=0
for i in range(1,n+1):
    if i%7==0 & i%9==0:
        total=total+i
print("Sum = ",total)
```

```python
n = int(input("Enter a number to add all prime numbers from 1 to n: "))
total = 0
for i in range(2, n+1):
    is_prime = True
    for j in range(2, int(i**0.5) + 1):
        if i % j == 0:
            is_prime = False
            break
    if is_prime:
        total += i
print("Sum of prime numbers is", total)
```

```python
n = int(input("Enter n: "))
sum = 0

for i in range(1, n + 1):
    if i % 2 != 0:
        sum = sum + i

print("Sum =", sum)
```

```python
n = int(input("Enter n: "))
sum = 0

for i in range(2, n + 1):
    prime = True

    for j in range(2, i):
        if i % j == 0:
            prime = False
            break

    if prime:
        sum = sum + i

print("Sum =", sum)
```
