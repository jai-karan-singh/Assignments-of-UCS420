```python
import pandas as pd

roll = 1024170199
d1 = 9
d2 = 9

categories = ["billing" , "account", "general"]

cat1 = categories[d1%3]
cat2 = categories[d2%3]

fixed_entries = [
    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price",
        "category": "billing"
    },
    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account"
    },
    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open",
        "category": "general"
    },
    {
        "question": "how can i pay the fee",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee",
        "category": "billing"
    }
]

personalized_entries = [
    {
        "question": "how can i pay my bill",
        "answer": "You can pay your bill using UPI, card, or net banking.",
        "keywords": "bill payment pay",
        "category": cat1
    },
    {
        "question": "how can i check my bill",
        "answer": "You can check your bill in the billing section.",
        "keywords": "bill check amount",
        "category": cat2
    }
]

entries = fixed_entries + personalized_entries
df = pd.DataFrame(entries)
print (df)

```

```python
def score_query(query , df):
    results = []
    query_words = query.lower().split()

    for i in range(len(df)):
        keywords = df.loc[i, "keywords"].lower().split()
        score = 0
        for word in query_words:
            if word in keywords:
                score = score+1

        if score > 0:
            results.append([df.loc[i, "question"], score])

    results.sort(key = lambda x:x[1], reverse = True)
    return results

query = input ("Enter a query: ")
print (score_query(query, df))

            
```

```python
def same_category(category_name, df):
    return df[df["category"] == category_name]["question"].tolist()

print ("Questions in category: ", cat1)
print (same_category(cat1,df))
```

```python
def score_query_with_ties (query, df):
    query_words = query.lower().split()
    scores = []
    for i in range(len(df)):
        keywords = df.loc[i, "keywords"].lower().split()
        score = 0

        for word in query_words:
            if word in keywords:
                score = score + 1
        scores.append(score)
    highest_score = max(scores)
    print ("Highest Score: ", highest_score)
    print ("Matching entries: ")

    for i in range(len(df)):
        if scores[i] == highest_score and highest_score > 0:
            print (df.loc[i, "question"])

print ("Tie demonstration: ")
score_query_with_ties("fee", df)
```
