```python
Roll_no = 1024170199
List = [int(d) for d in str(Roll_no)]
L = [x*10 for x in List]
print(L)

print("(ii) Using append() adding 9") #Will add 9 to the end of the list
L.append(9)
print(L)

print("(ii) Using insert() adding 1 at 10th index") #Will add 1 at the 10th index of the list
L.insert(10,1)
print(L)

print("(iii) Using remove() removing an element at first index")
L.remove(1)
print(L)

print("(iii) Using pop() removing an element from the end")
L.pop()
print(L)

print("(iv) Using sort() sorting list in ascending order")
L.sort()
print(L)

print("(iv) Using sort(reverse=True) sorting list in descending order")
L.sort(reverse=True)
print(L)

print("(v) Using slicing printing first three and last three elements")
print(L[0:3] + L[7:10])

print("(vi) Using list comprehension to create a new list containing only the elements of L that are greater than the average of L")
average = sum(L) / len(L)
new_L = [x for x in L if x > average]
print("List = ",L)
print("average of L = ",average)
print(new_L)
```

```python
scores = (90,90,70,40,20,10,10,10)
print ("Scores : ",scores)
high_score=max(scores)
high_index=scores.index(high_score)
low_score=min(scores)
low_index=scores.count(low_score)
print ("(i) Highest score in tuple and its index")
print (f"High score: {high_score} at index {high_index}")
print ("(i) Lowest score in tuple and number of time it occurs")
print (f"Score: {low_score} occurs {low_index} times")

reversed_scores = list(reversed(scores))
print ("(ii) Reversed scores: ",reversed_scores) # tuples can't be reversed in place because they are immutable, any change must create a new object

user_score = int(input("(iii) Enter a score to check index: "))
if user_score in scores:
    print (f"{user_score} found at index {scores.index(user_score)}")
else:
    print (f"{user_score} not found in the tuple")


try:
    scores[0] = 100
except TypeError as e:
    print(f"(iv) Error: {e}") #tuples are immutable, unlike lists which allow index assignment

first, second, *remaining = scores
print ("(v) ",first, second, remaining)
```

```python
import random
Roll_no = 1024170199
random.seed(Roll_no)

L = [random.randint(100,900) for _ in range(100)]
print ("(i) List: ",L)

odd_no = [num for num in L if num%2 != 0]
print(f"\n(ii) Odd numbers ({len(odd_no)}): {odd_no}")

even_no = [num for num in L if num%2 == 0]
print(f"\n(iii) Even numbers ({len(even_no)}): {even_no}")

def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

primes = [num for num in L if is_prime(num)]
print(f"\n(iv) Prime numbers ({len(primes)}): {primes}")

most_common = max(set(L), key=L.count)
print(f"\n(v) Most frequent number: {most_common}, occurs {L.count(most_common)} times")
```

```python
scores = (9,9,7,4,2,1,1,1)
A = { 7*digit for digit in scores}
B = { 9*digit for digit in scores}

print ("Set A: ",A)
print ("Set B: ",B)

union = A.union(B)
print ("(i) Union = ", union)

intersection = A.intersection(B)
print ("(ii) Intersection = ",intersection)

diff_A_B = A.difference(B)
print ("(iv) A - B = ",diff_A_B)
diff_B_A = B.difference(A)
print ("(iv) B - A = ",diff_B_A)
# difference() is one-directional only elements in the first set but not the second while symmetric_difference() combines both directions elements in either set but not both

symm_diff = A.symmetric_difference(B)
print ("(v) Symmetric Difference = ",symm_diff)

print(f"(vi) Is A a subset of B? {A.issubset(B)}")
print(f"(vi) Is B a superset of A? {B.issuperset(A)}")

X = int(input("(vii) Enter a value to remove from set A: "))
A.discard(X)  # discard() does not raise an error if X is absent unlike remove() which would throw a KeyError
print(f"\nSet A after discard: {A}")
```

```python
my_dict = {
    "name": "Jai Karan Singh",
    "roll_no": 1024170199,
    "branch": "COPC",
    "age": 20,
    "city": "Chandigarh"
}

my_dict["location"] = my_dict.pop("city")
print("(i) After renaming city -> location:", my_dict)


my_dict["cgpa"] = 8.4
print("(ii) After adding cgpa:", my_dict)


my_dict["age"] += 1
print("(iii) After incrementing age:", my_dict)


dict_pop = my_dict.copy()
removed_value = dict_pop.pop("branch")
print("(iv) After pop('branch'):", dict_pop)
print("pop() returned:", removed_value)

dict_del = my_dict.copy()
del dict_del["branch"]
print("(iv) After del dict_del['branch']:", dict_del)
# pop() removes the key and returns its value, while del removes the key but returns nothing


print("(v) Iterating with .items():")
for key, value in my_dict.items():
    print(f"{key} -> {value}")


if "email" in my_dict:
    print(my_dict["(vi) email"])
else:
    print("(vi) 'email' key not found in the dictionary")


friend_dict = {
    "name": "N Soma Sekhar",
    "roll_no": 1024170194,
    "branch": "COPC",
    "age": 21,
    "location": "Delhi"
}

merged_dict = {**my_dict, **friend_dict}
print("(vii) Merged dictionary:", merged_dict)
# when keys overlap, the values from the second dictionary (friend_dict) "win" and overwrite the first


string_only_dict = {k: v for k, v in my_dict.items() if isinstance(v, str)}
print("(viii) String-only key-value pairs:", string_only_dict)
```
