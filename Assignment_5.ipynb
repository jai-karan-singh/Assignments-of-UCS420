```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("Original array:", arr)
print("After adding 2:", arr + 2)
print("After multiplying by 3:", arr * 3)
print("After dividing by 2:", arr / 2)
```

```python
arr = np.array([1, 2, 3, 6, 4, 5])

print("Original array:", arr)
print("Reversed array:", arr[::-1])

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

values, counts = np.unique(x, return_counts=True)

most_frequent = values[np.argmax(counts)]

indices = np.where(x == most_frequent)[0]

print("Most frequent value:", most_frequent)
print("Indices:", indices)

y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

values, counts = np.unique(y, return_counts=True)

max_count = np.max(counts)

most_frequent = values[counts == max_count]

print("Most frequent values:", most_frequent)

for value in most_frequent:
    print(value, "indices:", np.where(y == value)[0])
```

```python
arr = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])

print("1st row, 2nd column:", arr[0, 1])
print("3rd row, 1st column:", arr[2, 0])
```

```python
Jai_Karan = np.linspace(10, 100, 25)

print("Array:")
print(Jai_Karan)

print("Dimensions:", Jai_Karan.ndim)
print("Shape:", Jai_Karan.shape)
print("Total elements:", Jai_Karan.size)
print("Data type:", Jai_Karan.dtype)
print("Total bytes:", Jai_Karan.nbytes)

row_array = Jai_Karan.reshape(1, 25)

print("\nRow array:")
print(row_array)

transpose_array = row_array.reshape(25, 1)

print("\nTranspose using reshape:")
print(transpose_array)

print("\nTranspose using T:")
print(row_array.T)
```

```python
ucs420_Jai_Karan = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original array:")
print(ucs420_Jai_Karan)

print("Mean:", np.mean(ucs420_Jai_Karan))
print("Median:", np.median(ucs420_Jai_Karan))
print("Maximum:", np.max(ucs420_Jai_Karan))
print("Minimum:", np.min(ucs420_Jai_Karan))
print("Unique elements:", np.unique(ucs420_Jai_Karan))

reshaped_ucs420_Jai_Karan = ucs420_Jai_Karan.reshape(4, 3)

print("Reshaped array:")
print(reshaped_ucs420_Jai_Karan)

resized_ucs420_Jai_Karan = np.resize(ucs420_Jai_Karan, (2, 3))

print("Resized array:")
print(resized_ucs420_Jai_Karan)
```
